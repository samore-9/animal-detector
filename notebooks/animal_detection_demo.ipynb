```python
from PIL import Image
from IPython.display import display
from app.detector import AnimalDetector

detector = AnimalDetector()   # first run downloads yolov8s.pt (needs internet)
print(detector.detector_name)
```

```python
IMAGE_PATH = 'sample_images/your_photo.jpg'   # <- put any photo in sample_images/ and change this

image = Image.open(IMAGE_PATH).convert('RGB')
result = detector.analyze(image, conf=0.35)

for a in result['animals']:
    print(f"{a['label']}: {a['count']} (best {a['best_confidence']:.0%})")
if not result['animals']:
    print('No animals found')
```

```python
# Draw the boxes on the image
from PIL import ImageDraw
vis = image.copy(); draw = ImageDraw.Draw(vis); W, H = vis.size
for d in result['detections']:
    if d['box']:
        x1, y1, x2, y2 = d['box'][0]*W, d['box'][1]*H, d['box'][2]*W, d['box'][3]*H
        draw.rectangle([x1, y1, x2, y2], outline='lime', width=4)
        draw.text((x1+4, y1+4), f"{d['label']} {d['confidence']:.0%}", fill='lime')
display(vis)
```
